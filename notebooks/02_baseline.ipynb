```python

import pandas as pd
import sys
sys.path.insert(0, '..')
from src.metrics import wape, bias

df = pd.read_csv('../data/processed/analysis_ready.csv', parse_dates=['date'])
```

```python

df = df.sort_values(['sku_id', 'date'])
df['baseline_pred'] = df.groupby('sku_id')['units_sold'].shift(364)
```

```python

holdout = df[df['date'] > df['date'].max() - pd.Timedelta(days=28)].dropna(subset=['baseline_pred'])
baseline_wape = wape(holdout['units_sold'], holdout['baseline_pred'])
print(f"Baseline WAPE (last 28 days holdout): {baseline_wape:.4f}")
```
