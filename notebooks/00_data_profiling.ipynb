```python

import pandas as pd
pd.set_option('display.width', 120)

sales = pd.read_csv('../data/raw/sales_daily.csv')
sku_master = pd.read_csv('../data/raw/sku_master.csv')
calendar = pd.read_csv('../data/raw/calendar.csv')
inventory = pd.read_csv('../data/raw/inventory_snapshots.csv')
```

```python

for name, df in [('sales_daily', sales), ('sku_master', sku_master),
                  ('calendar', calendar), ('inventory_snapshots', inventory)]:
    print(f"\n{'='*20} {name} {'='*20}")
    print("shape:", df.shape)
    print("dtypes:\n", df.dtypes)
    nulls = df.isnull().sum()
    print("nulls:\n", nulls[nulls > 0] if nulls.sum() else "none")
    print("duplicate rows:", df.duplicated().sum())
```

```python

sku_in_master = set(sku_master['SKU'])
sku_in_sales = set(sales['SKU'])
sku_in_inventory = set(inventory['SKU'])

print("sku_master SKUs:", len(sku_in_master))
print("sales_daily SKUs:", len(sku_in_sales))
print("inventory_snapshots SKUs:", len(sku_in_inventory))
print("\nsales_daily SKUs missing from sku_master:", len(sku_in_sales - sku_in_master))
print("inventory SKUs missing from sku_master:", len(sku_in_inventory - sku_in_master))
```

```python

print("negative units_sold:", (sales['Units_Sold'] < 0).sum())
print("negative price:", (sales['Price'] <= 0).sum())
implied_revenue = sales['Units_Sold'] * sales['Price']
mismatch = (implied_revenue - sales['Revenue']).abs() > 1.0
print("revenue mismatch rows:", mismatch.sum(), "/", len(sales))
```
